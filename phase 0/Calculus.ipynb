### GRADIENT DESCENT

```python
import numpy as np
```

```python
def quadratic(w):
    return w[0]**2 + 2*(w[1]**2) #x^2+2y^2

def derivative(w):
    return np.array([2*w[0],4*w[1]],dtype=np.float64)

def gradient(start_w,lr = 0.1,steps=40):
    w = np.array(start_w,dtype=np.float64)
    print(f"Starting Position : {w} , Initial Loss : {quadratic(w)}")
    for step in range(1, steps + 1):
        grad = derivative(w)
        w = w - lr * grad
        if step % 10 == 0:
            print(f"Step {step:02d} | Position: {w.round(4)} | Loss: {quadratic(w):.6f}")
    return w
```

```python
w_init = [5.0,-4.0]

final_w = gradient(w_init)
```
