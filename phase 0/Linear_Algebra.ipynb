### IMPLEMENT INVERSE USING GAUSS JORDAN ELIMINATION

```python
def inverse(A):
    A = A.astype(float)
    n = A.shape[0]

    M = np.column_stack((A,np.eye(n)))
    for i in range(n):
        M[i] = M[i]/M[i,i]

        
        for j in range(n):
            if i!=j:
                factor = M[j,i]
                M[j] = M[j] - factor*M[i]

    return M[:,n:]

```

```python
A = np.random.randint(0,40,size=16)
A = A.reshape(4,4)
A
```

```python
inverse(A)
```

```python
np.linalg.inv(A)
```
